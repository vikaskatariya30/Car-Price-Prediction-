```python
car=car.drop_duplicates().reset_index(drop=True)
```

```python
car.dropna(inplace=True)
car.reset_index(drop=True, inplace=True)
```

```python
car.shape
```

```python
car=car.drop(['mileage(km/ltr/kg)', 'engine', 'max_power', 'seats', 'year', 'km_driven', 'seller_type', 'owner'], axis=1)
```

```python
car['model']=car['name'].str.split().str.slice(0,3).str.join(' ')
```

```python
car.head()
```

```python
car['tags']=car['name']+" "+car['fuel']+" "+car['transmission']
```

```python
car['tags']=car['tags'].apply(lambda x:x.lower())
```

```python
from nltk.stem.porter import PorterStemmer
ps=PorterStemmer()
```

```python
def stem(text):
    y=[]
    for i in text.split():
        y.append(ps.stem(i))
    return " ".join(y)
```

```python
car['tags']=car['tags'].apply(stem)
```

```python
car.head()
```

```python
from sklearn.feature_extraction.text import CountVectorizer
cv=CountVectorizer(max_features=500, stop_words='english')
```

```python
vectors=cv.fit_transform(car['tags']).toarray()
```

```python
vectors
```

```python
list(cv.get_feature_names_out())
```

```python
from sklearn.preprocessing import MinMaxScaler
sc=MinMaxScaler()
price=sc.fit_transform(car[['selling_price']])
```

```python
final_vectors = np.hstack((vectors, price))
```

```python
from sklearn.metrics.pairwise import cosine_similarity
similarity = cosine_similarity(final_vectors)
print(similarity)
```

```python
def recommend(model):
    car_index=car[car['model']==model].index[0]
    distance=similarity[car_index]
    car_list=sorted(list(enumerate(distance)), reverse=True, key=lambda x:x[1])[1:6]
    for i in car_list:
        print(car.iloc[i[0]]['model'])
```

```python
recommend('Skoda Rapid 1.5')
```
